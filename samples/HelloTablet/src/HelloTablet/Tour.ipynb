# A tour of C# notebooks

Each cell runs on its own, and what one cell declares, the cells after it can use. Press **Run All** above, or put the cursor in a cell and press `Ctrl+Enter`.

- `Shift+Enter` runs a cell and moves to the next one
- A cell that ends with an expression shows its value
- Collections and objects show up as tables

```polyglot-notebook
var greeting = "Hello from a notebook";
greeting.ToUpperInvariant()
```

```polyglot-notebook
record Planet(string Name, double DistanceAu, int Moons);

var planets = new[]
{
    new Planet("Mercury", 0.39, 0),
    new Planet("Venus", 0.72, 0),
    new Planet("Earth", 1.0, 1),
    new Planet("Mars", 1.52, 2),
    new Planet("Jupiter", 5.2, 95),
};

planets.Where(p => p.Moons > 0)
```

## Showing values as you go

Call `display(...)` to show a value without ending the cell, and write to the console as usual.

```polyglot-notebook
foreach (var planet in planets.Take(2))
{
    display(planet);
}

Console.WriteLine($"{planets.Sum(p => p.Moons)} moons in total");
```

## Mistakes are fine

Misspell `planets` below and run the cell again: the error appears right under it, and the misspelled name is underlined.

```polyglot-notebook
planets.OrderByDescending(p => p.Moons).First().Name
```
